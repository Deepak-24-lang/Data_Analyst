```python
import numpy as np
```

```python
#NumPy Day 3: Reshaping, Sorting & 2D Arrays

sales = np.array([
    [1000, 1200, 1500],
    [2000, 1800, 2200],
    [1500, 1700, 1900]
])

print(sales)
print("dimensions:", sales.ndim)
print("shape:", sales.shape)
print("total elements:", sales.size)

# First row
print(sales[0])

# Second row
print(sales[1])

#printing the first column
print(sales[:, 0])

#second column
print(sales[:, 1])
```

```python
#Reshaping arrays
numbers = np.array([1, 2, 3, 4, 5, 6])

new_array  = numbers.reshape(2,3)
print(numbers.reshape(3, 2))
print(new_array)
```

```python
#Sorting arrays
prices = np.array([500, 200, 800, 300, 100])

print(np.sort(prices))  # Sorts the array in ascending order

#For descending order:
print(np.sort(prices)[::-1])  # Sorts the array in descending order
```

```python
#Find minimum, maximum and their positions
sales = np.array([2500, 4000, 1800, 5000, 3200])

print("Minimum sales:", np.min(sales))
print("Maximum sales:", np.max(sales))

print("Index of minimum sales:", np.argmin(sales))
print("Index of maximum sales:", np.argmax(sales))
```
