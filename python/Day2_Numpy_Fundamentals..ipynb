```python
import numpy as np
```

```python
#Filtering and Vectorized Operations

#Filtering arrays
sales = np.array([2500, 3000, 4500, 2000, 5000, 3500, 4000])

## Find sales greater than 3500
result = sales>3500
print(result)
```

```python
#Extract filtered values

highest_sales = sales[sales>3500]
print(highest_sales)
```

```python
#Use multiple conditions
selected_sales_and = sales[(sales>3000) & (sales<5000)]
print(selected_sales_and)

selected_sales_or = sales[(sales<3000) | (sales>4000)]
print(selected_sales_or)

selected_sales_not = sales[~(sales>3000)]
print(selected_sales_not)
```

```python
#Vectorized operations
prices = np.array([100, 200, 300, 400])

## Increase every price by 10%
new_prices = prices * 1.10
print(new_prices)
```

```python
numbers = np.array([10, 20, 30, 40])

print(numbers + 5)  # Add 5 to each element
print(numbers - 5)  # Subtract 5 from each element 
print(numbers * 2)  # Multiply each element by 2
print(numbers / 2)  # Divide each element by 2
```

```python
#Practical business example: Sales analysis

daily_sales = np.array([1200, 2500, 1800, 4000, 5200, 3000, 1500])

# Sales above 2000
print("High sales:", daily_sales[daily_sales > 2000])

# Count days with sales above 2000
print("Number of high-sales days:", np.sum(daily_sales > 2000))

# Add a 5% increase to all sales
increased_sales = daily_sales * 1.05

print("Increased sales:", increased_sales)
```

```python
marks = np.array([45, 78, 90, 32, 65, 88, 50])

high_marks = marks[marks > 60]
low_marks = marks[marks < 50]
count = marks[marks >= 75].size

print("High marks:", high_marks)
print("Low marks:", low_marks) 
print("count :", count) 
```

```python
#Vectorized operations
monthly_sales = np.array([10000, 12000, 15000, 18000, 20000])

increased_monthly_sales = monthly_sales * 1.10
deducted_monthly_sales = monthly_sales - (monthly_sales * 0.95)
total_sales = np.sum(monthly_sales)

print("Increased monthly sales:", increased_monthly_sales)
print("Deducted monthly sales:", deducted_monthly_sales)
print("Total sales:", total_sales)
```

```python
#Business analysis
product_prices = np.array([500, 1200, 800, 2000, 1500, 300])

High_price = product_prices[product_prices > 1000]
between_price = product_prices[(product_prices > 500) & (product_prices <= 1500)]
average_price = np.mean(product_prices)

print("High price products:", High_price)
print("Products priced between 500 and 1500:", between_price) 
print("Average product price:", average_price)
```
