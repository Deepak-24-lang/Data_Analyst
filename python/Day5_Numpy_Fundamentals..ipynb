```python
#Day 5: Advanced NumPy
import numpy as np
```

```python

#Combining NumPy arrays
week1 = np.array([1000, 2000, 3000])
week2 = np.array([4000, 5000, 6000])

combined = np.concatenate([week1, week2])
print(combined)
```

```python
#Understanding vstack() and hstack()
week1 = np.array([1000, 2000, 3000])
week2 = np.array([4000, 5000, 6000])

result = np.vstack((week1, week2))
result = np.hstack((week1, week2))

print(result)
```

```python
#Transpose an array
sales = np.array([
    [1000, 2000, 3000],
    [4000, 5000, 6000]
])

transposed_sales = sales.T
print(transposed_sales)

#Check the shape:
print(sales.shape)
print(transposed_sales.shape)

```

```python
#Advanced filtering
sales = np.array([1500, 2500, 3500, 4500, 5500, 6500])

#Sales greater than 3000
print(sales[sales > 3000])

#Sales between 2000 and 5000
print(sales[(sales > 2000) & (sales < 5000)])

#Sales greater than 5000 OR less than 2000
print(sales[(sales > 5000) | (sales < 2000)])

```

```python
#Replace values using conditions
sales = np.array([1500, 2500, 3500, 4500, 5500, 6500])
sales_status = np.where(sales < 3000, "low", "high")

print(sales_status)
```

```python
#Business KPIs using NumPy
product_sales = np.array([10000, 15000, 8000, 20000])

print("Total sales:", np.sum(product_sales))
print("Average sales:", np.mean(product_sales))
print("Highest sales:", np.max(product_sales))
print("Lowest sales:", np.min(product_sales))
print("Number of products:", product_sales.size)

#Calculate contribution percentage
contribution_percentage = (product_sales / np.sum(product_sales)) * 100
print("Contribution percentage:", contribution_percentage)
```

```python
#Mini project: Retail sales analysis
sales_data = np.array([
    [10000, 12000, 15000],
    [8000, 9000, 11000],
    [15000, 14000, 16000],
    [12000, 13000, 18000]
])
```

```python
#Inspect the dataset
print("sales_data :", sales_data)
print("Shape of sales_data:", sales_data.shape)
print("dimensions of sales_data:", sales_data.ndim)
print("size of sales_data:", sales_data.size)
```

```python
#Calculate total sales for each product
print("Total sales for each product:", np.sum(sales_data, axis=1))
```

```python
#Calculate total sales for each month
print("Total sales for each month:", np.sum(sales_data, axis=0))
```
