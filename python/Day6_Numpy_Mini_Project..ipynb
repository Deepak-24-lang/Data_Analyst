```python
import numpy as np
```

```python
#Create the dataset
sales = np.array([
    [10000, 12000, 15000],
    [8000, 9000, 11000],
    [15000, 14000, 16000],
    [12000, 13000, 18000]
])

print(sales)
print("Shape:", sales.shape)
```

```python
#Calculate total sales per product
product_sales = np.sum(sales, axis=1)

print(product_sales)
```

```python
#Calculate monthly sales
monthly_sales = np.sum(sales, axis=0)

print(monthly_sales)
```

```python
#Find the best-performing product
best_product_index = np.argmax(product_sales)

products = np.array(["Laptop", "Mobile", "Tablet", "Monitor"])

print("Best product:", products[best_product_index])
print("Sales:", product_sales[best_product_index])
```

```python
#Find the best month
months = np.array(["January", "February", "March"])

best_month_index = np.argmax(monthly_sales)

print("Best month:", months[best_month_index])
print("Sales:", monthly_sales[best_month_index])
```

```python
#Calculate business KPIs
total_sales = np.sum(sales)
average_sales = np.mean(sales)
highest_sale = np.max(sales)
lowest_sale = np.min(sales)

print("Total Sales:", total_sales)
print("Average Sale:", average_sales)
print("Highest Sale:", highest_sale)
print("Lowest Sale:", lowest_sale)
```
