```python
import numpy as np
```

```python
#NumPy provides a random number generator.
rng = np.random.default_rng()
numbers = rng.integers(1, 10, size=(5))
print(numbers)
```

```python
#Sometimes you want the same random numbers every time you run your code.
rng = np.random.default_rng(42)
numbers = rng.integers(1,10, size=(5))
print(numbers)
```

```python
#Generate decimal numbers
Decimal = np.random.default_rng(42)
output = Decimal.random(5)
print(output)
```

```python
#Select random values from an array
rng = np.random.default_rng(42)

Products = np.array([100,200,300,400,500])
selected_products = rng.choice(Products, size=(3), replace=False)
print(selected_products)
```

```python
#Generate normally distributed values
rng = np.random.default_rng(42)

data = rng.normal(loc=100, scale=15, size=(5))
print(data)
```

```python
#Understanding missing values
sales = np.array([1200, 1500, np.nan, 1800, 2000])

#Check for missing values
missing_values = np.isnan(sales)
print(missing_values)

#Count missing values
num_missing = np.sum(missing_values)
print(num_missing)

```

```python
#Calculate statistics while ignoring missing values
sales = np.array([1200, 1500, np.nan, 1800, 2000])

print("Normal mean:", np.mean(sales))
print("Mean ignoring NaN:", np.nanmean(sales))
print("Total:", np.nansum(sales))
print("Average:", np.nanmean(sales))
print("Minimum:", np.nanmin(sales))
print("Maximum:", np.nanmax(sales))
```

```python
#Replacing missing values
sales = np.array([1200, 1500, np.nan, 1800, 2000])

average_sales = np.nanmean(sales)

clean_sales = np.where(
    np.isnan(sales),
    average_sales,
    sales
)

print(clean_sales)
```

```python
weekly_sales = np.array([
    2500,
    3000,
    np.nan,
    2000,
    5000,
    np.nan,
    4000
])
```

```python
#Inspect the dataset
print("sales data:", weekly_sales)
print("Count of sales data :",np.size(weekly_sales))
print("Count of missing values:", np.sum(np.isnan(weekly_sales)))
```

```python
#Calculate total and average sales
total_sales = np.nansum(weekly_sales)
average_sales = np.nanmean(weekly_sales)

print("Total sales:", total_sales)
print("Average sales:", average_sales)
```

```python
#Find the highest and lowest sales
print("Highest sales:", np.nanmax(weekly_sales))
print("Lowest sales:", np.nanmin(weekly_sales))
```

```python
#Find days with sales greater than 3000
high_sales_days = weekly_sales[weekly_sales > 3000]
print("Days with sales greater than 3000:", high_sales_days)
```

```python
#Fill missing values with the average
average_sales = np.nanmean(weekly_sales)
filled_sales = np.where(
    np.isnan(weekly_sales),
    average_sales,
    weekly_sales
)

print("Filled sales data:", filled_sales)
```

```python
#Compare original and cleaned data
print("Original sales data:", weekly_sales)
print("Cleaned sales data:", filled_sales)

print("Original total:", np.nansum(weekly_sales))
print("Cleaned total:", np.sum(clean_sales))
```
