```python
#Understand array dimensions
#One-dimensional array (1D)
a =np.array([10, 20, 30, 40, 50])

print(a)
print(a.ndim)
```

```python
#Two-dimensional array (2D)
b = np.array([[1, 2, 3],
               [4, 5, 6]
            ])

print(b)
print(b.ndim)
print(b.shape)
```

```python
#arrays using NumPy functions 
np.zeros(5)
np.ones(5)
np.arange(1,11)
np.arange(0,101,10)
```

```python
#Indexing: Access individual values
sales=np.array([1000,2000,3000,4000,5000])

print(sales[0])
print(sales[3])
print(sales[-1])
```

```python
#Slicing: Access multiple values
sales = np.array([1000, 2000, 3000, 4000, 5000])
print(sales[1:4])
```

```python
#Perform calculations on arrays 
sales = np.array([1000, 2000, 3000, 4000, 5000])

print(np.sum(sales))
print(np.mean(sales))
print(np.min(sales))
print(np.max(sales))
print(np.median(sales))
```

```python
weekly_sales = np.array([2500, 3000, 4500, 2000, 5000, 3500, 4000])

print("Sales array :",weekly_sales)
print("Total weekly sales:", np.sum(weekly_sales))
print("Average weekly sales:", np.mean(weekly_sales))
print("maximum weekly sales:", np.max(weekly_sales))
print("minimum weekly sales:", np.min(weekly_sales))
print("first three days sales:", weekly_sales[0:3])
print("last two days sales:", weekly_sales[-2:])
print("no of days",np.size(weekly_sales))
```
